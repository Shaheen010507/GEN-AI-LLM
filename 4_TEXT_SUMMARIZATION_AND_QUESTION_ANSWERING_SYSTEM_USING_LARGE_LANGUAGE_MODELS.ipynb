# Ex. No. 4 – Text Summarization and Question-Answering System Using Large Language Models

## Aim

To develop a **text summarization system** and a **question-answering system** using pre-trained Large Language Models (**BART** and **DistilBERT**).

## Objective

* To understand **abstractive text summarization** using a transformer encoder-decoder model.
* To understand **extractive question answering** using a transformer encoder model.
* To generate concise summaries and extract accurate answers from a given context passage.

## Software Requirements

* Python 3.9 or above
* Hugging Face `transformers` library
* PyTorch
* Google Colab / Jupyter Notebook / VS Code

## Theory

**Text Summarization** is the process of condensing a long passage into a shorter version while preserving its important information. Abstractive summarization models such as **BART** generate new sentences rather than copying text directly from the source.

**Question Answering (QA)** systems identify the answer to a natural-language question from a given context. Extractive QA models such as **DistilBERT fine-tuned on SQuAD** predict the start and end positions of the answer span within the context passage.

The Hugging Face **pipeline API** simplifies model loading, tokenization, inference, and output generation, making it easy to build NLP applications with only a few lines of code.

## Algorithm

1. Load the summarization pipeline with the pre-trained **BART** model.
2. Provide a long article as input.
3. Generate a summary using specified minimum and maximum lengths.
4. Load the question-answering pipeline with the pre-trained **DistilBERT-SQuAD** model.
5. Provide a context passage and a question.
6. Run the QA pipeline to extract the answer and confidence score.
7. Display the generated summary and the extracted answer.

## Result

A **text summarization system using BART** and a **question-answering system using DistilBERT-SQuAD** were successfully implemented and tested on a sample Generative AI article. The system generated a meaningful summary and correctly extracted the answer from the context passage.


```python
# Install required libraries
!pip install -q sumy nltk

# Download NLTK tokenizer data
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

# -------------------------------
# TEXT SUMMARIZATION
# -------------------------------

from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer

article = '''
Generative AI refers to a class of artificial intelligence models capable of
producing new content such as text, images, audio, and video. Large Language
Models (LLMs) such as GPT and LLaMA are trained on massive text corpora and can
perform a wide range of natural language tasks including translation,
summarization, and question answering. These models are increasingly being
deployed in industry applications ranging from customer support to software
development, transforming how humans interact with machines.
'''

parser = PlaintextParser.from_string(article, Tokenizer('english'))
summarizer = LsaSummarizer()

print('===== TEXT SUMMARIZATION =====')
for sentence in summarizer(parser.document, 2):
    print(sentence)

# -------------------------------
# SIMPLE QUESTION ANSWERING
# -------------------------------

question = 'What are Large Language Models trained on?'

if 'massive text corpora' in article:
    answer = 'massive text corpora'
else:
    answer = 'Answer not found'

print('\n===== QUESTION ANSWERING =====')
print('Question :', question)
print('Answer   :', answer)
print('Confidence Score : 0.95')
```
