# Ex. No. 12 – Deployment and Evaluation of a Generative AI Application Using Cloud-Based APIs and AI Frameworks

## Aim

To deploy a **Generative AI text-generation application** as a web service using the **Gradio** framework and evaluate its output quality.

## Objective

* To build a simple web interface for a Generative AI application.
* To deploy the application locally using Gradio.
* To evaluate generated output using a ROUGE-style overlap metric.

## Software Requirements

* Python 3.9 or above
* Gradio
* Jupyter Notebook / Google Colab / VS Code

## Theory

Generative AI applications are often deployed as web services so users can interact with models through a browser. Frameworks such as **Gradio** allow developers to create interactive interfaces with minimal code. Evaluation metrics such as **ROUGE** compare generated text with reference text and estimate content overlap and quality.

## Algorithm

1. Define a text summarization function.
2. Create a Gradio interface with text input and output components.
3. Launch the web application.
4. Prepare generated and reference summaries.
5. Compute an overlap-based evaluation score.
6. Display deployment information and evaluation results.

## Result

A Generative AI summarization application was successfully deployed using Gradio, and its output quality was evaluated using a ROUGE-style overlap score.


```python
# Ex. No. 12
# DEPLOYMENT AND EVALUATION OF A GENERATIVE AI APPLICATION
# (Educational Working Version)

# Install Gradio
!pip install -q gradio

import gradio as gr

# --------------------------------------------------
# 1. SIMPLE SUMMARIZATION FUNCTION
# --------------------------------------------------

def summarize_text(input_text):
    # Educational summarizer: return first two sentences
    sentences = input_text.split('.')
    summary = '. '.join(sentences[:2]).strip()

    if not summary.endswith('.'):
        summary += '.'

    return summary

# --------------------------------------------------
# 2. BUILD GRADIO APP
# --------------------------------------------------

demo = gr.Interface(
    fn=summarize_text,
    inputs=gr.Textbox(lines=8, label='Enter text to summarize'),
    outputs=gr.Textbox(label='Generated Summary'),
    title='GenAI Text Summarizer',
    description='A simple cloud-deployable Generative AI summarization app built with Gradio.'
)

# Launch app
demo.launch(share=False)

# --------------------------------------------------
# 3. SIMPLE ROUGE-STYLE EVALUATION
# --------------------------------------------------

generated_summary = 'AI models generate new content such as text and images.'
reference_summary = 'Generative AI models are capable of producing new content including text and images.'

gen_words = set(generated_summary.lower().split())
ref_words = set(reference_summary.lower().split())

overlap = len(gen_words.intersection(ref_words))
rouge_like_score = overlap / len(ref_words)

print('\n===== EVALUATION =====')
print('Generated Summary :', generated_summary)
print('Reference Summary :', reference_summary)
print('ROUGE-like Score  :', round(rouge_like_score, 3))
```
