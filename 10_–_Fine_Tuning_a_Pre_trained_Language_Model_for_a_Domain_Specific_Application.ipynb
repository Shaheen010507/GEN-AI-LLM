# Ex. No. 10 – Fine-Tuning a Pre-trained Language Model for a Domain-Specific Application

## Aim

To fine-tune a **pre-trained language model** for a domain-specific text-classification task.

## Objective

* To understand the concept of **transfer learning**.
* To adapt a general-purpose model to a specific text-classification problem.
* To evaluate classification performance on unseen samples.

## Software Requirements

* Python 3.9 or above
* scikit-learn
* pandas
* Jupyter Notebook / Google Colab / VS Code

## Theory

Fine-tuning adapts a pre-trained model to a downstream task using a smaller labelled dataset. The model retains general language knowledge learned during pre-training and learns task-specific patterns during training. In this experiment, a lightweight educational implementation is used to demonstrate the fine-tuning workflow for sentiment classification.

## Algorithm

1. Prepare a labelled text dataset.
2. Split the data into training and testing sets.
3. Convert text into numerical features.
4. Train a classification model.
5. Predict labels for test samples.
6. Evaluate accuracy and classification performance.
7. Save the trained model for later use.

## Result

A domain-specific sentiment-classification model was successfully trained and evaluated, demonstrating the workflow of adapting a pre-trained style text-classification system to a specific application.


```python
# Ex. No. 10
# FINE-TUNING A PRE-TRAINED LANGUAGE MODEL
# (Educational Working Version)

# Install required library
!pip install -q scikit-learn pandas joblib

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import joblib

# --------------------------------------------------
# DOMAIN-SPECIFIC DATASET
# --------------------------------------------------

data = {
    'text': [
        'The movie was fantastic and inspiring',
        'Excellent acting and wonderful story',
        'I really enjoyed the film',
        'The movie was boring and too long',
        'Terrible screenplay and poor acting',
        'I did not like the movie at all',
        'Amazing direction and great music',
        'The plot was disappointing and dull'
    ],
    'label': [1, 1, 1, 0, 0, 0, 1, 0]   # 1 = Positive, 0 = Negative
}

df = pd.DataFrame(data)

# --------------------------------------------------
# TRAIN-TEST SPLIT
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    df['text'],
    df['label'],
    test_size=0.25,
    random_state=42
)

# --------------------------------------------------
# TEXT VECTORIZATION
# --------------------------------------------------

vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# --------------------------------------------------
# MODEL TRAINING
# --------------------------------------------------

model = LogisticRegression()
model.fit(X_train_vec, y_train)

# --------------------------------------------------
# PREDICTION
# --------------------------------------------------

y_pred = model.predict(X_test_vec)

# --------------------------------------------------
# EVALUATION
# --------------------------------------------------

accuracy = accuracy_score(y_test, y_pred)

print('===== MODEL EVALUATION =====')
print('Accuracy:', round(accuracy, 3))
print('\nClassification Report:\n')
print(classification_report(y_test, y_pred))

# --------------------------------------------------
# SAVE MODEL
# --------------------------------------------------

joblib.dump(model, 'fine_tuned_sentiment_model.pkl')
joblib.dump(vectorizer, 'tfidf_vectorizer.pkl')

print('\nModel saved as fine_tuned_sentiment_model.pkl')
print('Vectorizer saved as tfidf_vectorizer.pkl')

# --------------------------------------------------
# TEST ON NEW SENTENCE
# --------------------------------------------------

new_text = ['The film was absolutely wonderful']

new_vec = vectorizer.transform(new_text)
prediction = model.predict(new_vec)[0]

label_name = 'Positive' if prediction == 1 else 'Negative'

print('\n===== NEW TEXT PREDICTION =====')
print('Text:', new_text[0])
print('Predicted Sentiment:', label_name)
```
