# Ex. No. 1 – Text Generation Using Pre-trained Foundation Models

## Aim

To develop a text generation application using a pre-trained foundation model (**GPT-2**) with the Hugging Face **Transformers** library.

## Objective

* To understand how pre-trained foundation models generate coherent text from a given prompt.
* To explore decoding strategies such as **greedy search, sampling, top-k, and top-p (nucleus) sampling**.
* To analyse the effect of generation parameters on output quality and diversity.

## Technologies Used

* Python 3.9+
* Hugging Face Transformers
* PyTorch
* Google Colab / Jupyter Notebook

## Experiment Summary

In this experiment, a pre-trained **GPT-2** model was loaded using the Hugging Face `pipeline` API. A text prompt was provided, and multiple text sequences were generated using sampling-based decoding with parameters such as `temperature`, `top_k`, and `top_p`. The generated outputs demonstrated coherent continuation of the input prompt and showed how decoding strategies influence creativity and diversity.

## Learning Outcomes

* Understanding of **foundation models** and **transformer-based text generation**.
* Hands-on experience with the Hugging Face `pipeline` API.
* Practical knowledge of **sampling techniques** used in large language models.
* Ability to generate and evaluate AI-generated text for different prompts.

## Result

A text generation application using the pre-trained **GPT-2 foundation model** was successfully implemented, and coherent text was generated from the given prompt using sampling-based decoding strategies.


```python
from transformers import pipeline, set_seed

# Load the pre-trained GPT-2 text generation pipeline
generator = pipeline("text-generation", model="gpt2")

set_seed(42)

# Input prompt
prompt = "Artificial Intelligence will transform the future of"

# Generate text
outputs = generator(
    prompt,
    max_length=60,
    num_return_sequences=2,
    temperature=0.8,
    top_k=50,
    top_p=0.95,
    do_sample=True
)

# Display generated texts
for i, out in enumerate(outputs, 1):
    print(f"--- Generated Text {i} ---")
    print(out["generated_text"])
    print()
```
