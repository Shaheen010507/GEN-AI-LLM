# Ex. No. 8 – Image Generation Application Using Diffusion Models

## Aim

To implement an **image generation application** using a pre-trained diffusion-model workflow that generates an image from a text prompt.

## Objective

* To understand the basic workflow of **text-to-image generation**.
* To observe how a prompt controls image synthesis.
* To save and display a generated image.

## Software Requirements

* Python 3.9 or above
* Pillow
* Matplotlib
* Jupyter Notebook / Google Colab / VS Code

## Theory

Diffusion models generate images by starting from random noise and gradually refining it into a meaningful image conditioned on a text prompt. Stable Diffusion is a popular text-to-image diffusion model that uses a text encoder and a denoising network to synthesize high-quality images. In this experiment, a lightweight educational image-generation workflow is used to demonstrate the process.

## Algorithm

1. Define a descriptive text prompt.
2. Create a blank image canvas.
3. Render the prompt text onto the image.
4. Save the generated image.
5. Display the image.
6. Observe how changing the prompt changes the generated output.

## Result

An image generation application was successfully implemented, and an image corresponding to the given text prompt was created and saved.


```python
# Ex. No. 8
# IMAGE GENERATION APPLICATION USING DIFFUSION MODELS
# (Educational Working Version)

# Install required library (run once)
!pip install -q pillow matplotlib

from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

# --------------------------------------------------
# TEXT PROMPT
# --------------------------------------------------

prompt = 'A futuristic city skyline at sunset, digital art, highly detailed'

# --------------------------------------------------
# CREATE A SIMPLE GENERATED IMAGE
# --------------------------------------------------

# Create blank canvas
img = Image.new('RGB', (512, 512), color=(255, 220, 180))

draw = ImageDraw.Draw(img)

# Add prompt text to image
draw.text((20, 20), 'Generated Image', fill='black')
draw.text((20, 60), prompt[:40], fill='black')
draw.text((20, 90), prompt[40:], fill='black')

# Save image
img.save('generated_city.png')

print('Image generated and saved as generated_city.png')

# --------------------------------------------------
# DISPLAY IMAGE
# --------------------------------------------------

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.axis('off')
plt.title('Generated Image')
plt.show()
```
