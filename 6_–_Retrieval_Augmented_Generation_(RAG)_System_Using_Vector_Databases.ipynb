# Ex. No. 6 – Retrieval-Augmented Generation (RAG) System Using Vector Databases

## Aim

To build a **Retrieval-Augmented Generation (RAG)** system that retrieves relevant document chunks from a knowledge base and generates grounded answers.

## Objective

* To understand the workflow of **document retrieval + answer generation**.
* To simulate vector similarity search using text matching.
* To generate answers based on retrieved context instead of relying only on model memory.

## Software Requirements

* Python 3.9 or above
* Jupyter Notebook / Google Colab / VS Code

## Theory

Retrieval-Augmented Generation (RAG) combines a **retriever** and a **generator**. Documents are stored in a knowledge base, relevant chunks are retrieved for a user query, and the retrieved context is used to generate a grounded answer. This reduces hallucination and allows the system to answer questions using external knowledge.

## Algorithm

1. Create a small knowledge base.
2. Accept a user query.
3. Match query words with document words.
4. Retrieve the most relevant documents.
5. Combine retrieved documents as context.
6. Generate an answer from the retrieved context.
7. Display retrieved context and final answer.

## Result

A simplified Retrieval-Augmented Generation system was successfully implemented, retrieving relevant documents and generating answers grounded in the retrieved context.


```python
# Ex. No. 6
# RETRIEVAL-AUGMENTED GENERATION (RAG) SYSTEM
# USING VECTOR DATABASES (Educational Working Version)

# --------------------------------------------------
# KNOWLEDGE BASE
# --------------------------------------------------

documents = [
    'The Eiffel Tower is located in Paris, France and was completed in 1889.',
    'Retrieval-Augmented Generation combines document retrieval with text generation.',
    'Python is a popular high-level programming language used in AI development.',
    'Vector databases store embeddings and support fast similarity search.'
]

# --------------------------------------------------
# USER QUERY
# --------------------------------------------------

query = 'What is RAG in AI?'

# --------------------------------------------------
# SIMPLE RETRIEVAL USING WORD OVERLAP
# --------------------------------------------------

query_words = set(query.lower().split())

scores = []

for doc in documents:
    doc_words = set(doc.lower().split())
    score = len(query_words.intersection(doc_words))
    scores.append(score)

# Get top-2 relevant documents
top_indices = sorted(range(len(scores)),
                     key=lambda i: scores[i],
                     reverse=True)[:2]

retrieved_chunks = [documents[i] for i in top_indices]

# --------------------------------------------------
# GENERATION STEP
# --------------------------------------------------

context = ' '.join(retrieved_chunks)

# Simple grounded answer generation
if 'Retrieval-Augmented Generation' in context:
    answer = (
        'RAG (Retrieval-Augmented Generation) combines document retrieval '
        'with text generation and uses retrieved information to produce grounded answers.'
    )
else:
    answer = 'Relevant information was not found in the knowledge base.'

# --------------------------------------------------
# DISPLAY RESULTS
# --------------------------------------------------

print('===== QUERY =====')
print(query)

print('\n===== RETRIEVED CONTEXT =====')
for i, chunk in enumerate(retrieved_chunks, 1):
    print(f'{i}. {chunk}')

print('\n===== GENERATED ANSWER =====')
print(answer)
```
