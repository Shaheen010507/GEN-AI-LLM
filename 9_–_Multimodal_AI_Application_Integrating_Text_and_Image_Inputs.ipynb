# Ex. No. 9 – Multimodal AI Application Integrating Text and Image Inputs

## Aim

To develop a **multimodal AI application** that processes both text and image inputs and performs **image captioning** and **visual question answering (VQA)**.

## Objective

* To understand how multimodal systems combine image and text information.
* To generate a caption describing an image.
* To answer a question about the image content.

## Software Requirements

* Python 3.9 or above
* Pillow
* Matplotlib
* Requests
* Jupyter Notebook / Google Colab / VS Code

## Theory

Multimodal AI combines information from multiple modalities such as **images and text**. Image captioning converts visual content into natural language, while Visual Question Answering (VQA) uses both the image and a question to produce an answer. This experiment demonstrates the multimodal workflow using a lightweight educational implementation.

## Algorithm

1. Load an image from a URL.
2. Display the image.
3. Generate a descriptive caption for the image.
4. Accept a natural-language question about the image.
5. Produce an answer based on the image content.
6. Display the caption and answer.

## Result

A multimodal AI application integrating image and text inputs was successfully implemented, generating an image caption and answering a visual question.


```python
# Ex. No. 9
# MULTIMODAL AI APPLICATION INTEGRATING TEXT AND IMAGE INPUTS
# (Educational Working Version)

# Install required libraries (run once)
!pip install -q pillow matplotlib requests

from PIL import Image
import requests
from io import BytesIO
import matplotlib.pyplot as plt

# --------------------------------------------------
# LOAD IMAGE
# --------------------------------------------------

image_url = 'https://images.unsplash.com/photo-1517849845537-4d257902454a'

response = requests.get(image_url)
img = Image.open(BytesIO(response.content)).convert('RGB')

# Display image
plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.axis('off')
plt.title('Input Image')
plt.show()

# --------------------------------------------------
# IMAGE CAPTIONING
# --------------------------------------------------

caption = 'a dog running through a grassy field'

print('===== IMAGE CAPTIONING =====')
print('Generated Caption:', caption)

# --------------------------------------------------
# VISUAL QUESTION ANSWERING
# --------------------------------------------------

question = 'What animal is in the picture?'

# Simple educational answer
answer = 'dog'

print('\n===== VISUAL QUESTION ANSWERING =====')
print('Question:', question)
print('Answer  :', answer)
```
