# Ex. No. 7 – AI-Powered Code Generation and Debugging Assistant

## Aim

To develop an **AI-powered assistant** that generates code from natural-language instructions and helps identify and fix bugs in existing code.

## Objective

* To understand how AI models can generate source code from text instructions.
* To simulate code generation and debugging assistance.
* To compare buggy and corrected code snippets.

## Software Requirements

* Python 3.9 or above
* Jupyter Notebook / Google Colab / VS Code

## Theory

AI code assistants use transformer models trained on large code corpora to generate code, explain code, and suggest fixes for programming errors. In this experiment, a lightweight rule-based implementation is used to demonstrate the workflow of **code generation** and **bug fixing**.

## Algorithm

1. Accept a natural-language programming instruction.
2. Generate a suitable Python function template.
3. Provide a buggy code snippet.
4. Analyse the bug and produce a corrected version.
5. Display the generated code and debugging suggestion.
6. Compare buggy and fixed code outputs.

## Result

An AI-powered code generation and debugging assistant was successfully implemented, generating Python code from a natural-language instruction and suggesting a corrected version for buggy code.


```python
# Ex. No. 7
# AI-POWERED CODE GENERATION AND DEBUGGING ASSISTANT
# (Educational Working Version)

# --------------------------------------------------
# SIMPLE CODE GENERATION FUNCTION
# --------------------------------------------------

def generate_code(prompt):
    prompt = prompt.lower()

    if 'prime' in prompt:
        return '''
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
'''

    elif 'factorial' in prompt:
        return '''
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
'''

    else:
        return '# No code template available for this instruction.'

# --------------------------------------------------
# 1. CODE GENERATION
# --------------------------------------------------

instruction = 'Write a Python function to check if a number is prime'

generated = generate_code(instruction)

print('===== GENERATED CODE =====')
print('Instruction:', instruction)
print(generated)

# --------------------------------------------------
# 2. DEBUGGING ASSISTANT
# --------------------------------------------------

buggy_code = '''
def factorial(n):
    result = 0
    for i in range(1, n + 1):
        result = result * i
    return result
'''

fixed_code = '''
def factorial_fixed(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result
'''

print('===== BUGGY CODE =====')
print(buggy_code)

print('===== FIXED CODE =====')
print(fixed_code)

# --------------------------------------------------
# VERIFY FIX
# --------------------------------------------------

def factorial_fixed(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result

print('===== VERIFICATION =====')
print('factorial_fixed(5) =', factorial_fixed(5))
```
