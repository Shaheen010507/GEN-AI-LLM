# Ex. No. 11 – AI-Based Content Generation System for Text, Image and Multimedia Applications

## Aim

To develop an **AI-based multimedia content generation system** that produces **text, image, and audio outputs** from a single user topic.

## Objective

* To understand multimedia content generation pipelines.
* To generate textual content from a topic.
* To create a related image.
* To generate an audio narration of the generated text.

## Software Requirements

* Python 3.9 or above
* Pillow
* Matplotlib
* pyttsx3
* Jupyter Notebook / Google Colab / VS Code

## Theory

Modern AI systems often combine multiple generative components to create multimedia content. A text generator creates textual content, an image generator produces a visual representation, and a text-to-speech system converts the generated text into audio. This experiment demonstrates an end-to-end multimedia workflow using lightweight educational components.

## Algorithm

1. Accept a topic from the user.
2. Generate a short paragraph about the topic.
3. Create a simple image representing the topic.
4. Convert the generated text into speech.
5. Save the text, image, and audio outputs.
6. Display the generated multimedia content.

## Result

An AI-based multimedia content generation system was successfully implemented, producing text, image, and audio outputs from a single topic input.


```python
# Ex. No. 11
# AI-BASED CONTENT GENERATION SYSTEM FOR TEXT, IMAGE AND MULTIMEDIA APPLICATIONS
# (Educational Working Version)

# Install required libraries (run once)
!pip install -q pillow matplotlib pyttsx3

from PIL import Image, ImageDraw
import matplotlib.pyplot as plt
import pyttsx3

# --------------------------------------------------
# USER TOPIC
# --------------------------------------------------

topic = 'The benefits of renewable energy'

# --------------------------------------------------
# 1. TEXT GENERATION
# --------------------------------------------------

generated_text = (
    'Renewable energy sources such as solar and wind power reduce carbon emissions, '
    'improve environmental sustainability, and help create a cleaner future for '
    'coming generations.'
)

print('===== GENERATED TEXT =====')
print(generated_text)

# Save text
with open('content_text.txt', 'w') as f:
    f.write(generated_text)

# --------------------------------------------------
# 2. IMAGE GENERATION
# --------------------------------------------------

img = Image.new('RGB', (512, 512), color=(220, 255, 220))
draw = ImageDraw.Draw(img)

draw.text((20, 20), 'Generated Image', fill='black')
draw.text((20, 70), topic, fill='black')

img.save('content_image.png')

print('\nImage saved as content_image.png')

# Display image
plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.axis('off')
plt.title('Generated Image')
plt.show()


# --------------------------------------------------
# 3. AUDIO GENERATION (Working Version)
# --------------------------------------------------

!pip install -q gTTS

from gtts import gTTS

tts = gTTS(text=generated_text, lang='en')
tts.save('content_audio.mp3')

print('Audio saved as content_audio.mp3')
# --------------------------------------------------
# FINAL OUTPUT FILES
# --------------------------------------------------

print('\n===== GENERATED FILES =====')
print('1. content_text.txt')
print('2. content_image.png')
print('3. content_audio.mp3')
```
