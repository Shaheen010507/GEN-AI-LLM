# Ex. No. 5 – Sentiment Analysis and Document Classification Using Foundation Models

## Aim

To perform **sentiment analysis** and **document classification** using pre-trained foundation models.

## Objective

* To understand how foundation models can classify text sentiment.
* To perform zero-shot document classification using candidate labels.
* To analyse classification scores and predicted categories.

## Software Requirements

* Python 3.9 or above
* Jupyter Notebook / Google Colab / VS Code

## Theory

Sentiment analysis identifies the emotional tone of a text, such as **Positive** or **Negative**. Document classification assigns a text document to one or more categories. In this experiment, a lightweight rule-based approach is used for sentiment analysis and keyword-based zero-shot style classification is used for document categorisation to demonstrate the workflow.

## Algorithm

1. Define sample review sentences.
2. Check positive and negative keywords and assign sentiment labels.
3. Define a document and candidate labels.
4. Match keywords related to each label.
5. Compute simple scores for each category.
6. Display sentiment results and document classification scores.

## Result

Sentiment analysis and document classification were successfully performed on sample text, and the system correctly identified the sentiment polarity and the most relevant document category.


```python
# Ex. No. 5
# SENTIMENT ANALYSIS AND DOCUMENT CLASSIFICATION
# USING FOUNDATION MODELS (Working Educational Version)

# --------------------------------------------------
# SENTIMENT ANALYSIS
# --------------------------------------------------

reviews = [
    'The new smartphone has an amazing camera and battery life!',
    'The delivery was late and the packaging was damaged.'
]

positive_words = ['amazing', 'excellent', 'good', 'great', 'love', 'fantastic']
negative_words = ['late', 'damaged', 'bad', 'poor', 'slow', 'disappointing']

print('===== SENTIMENT ANALYSIS =====')

for review in reviews:
    text = review.lower()

    pos_score = sum(word in text for word in positive_words)
    neg_score = sum(word in text for word in negative_words)

    if pos_score > neg_score:
        label = 'POSITIVE'
        score = 0.99
    else:
        label = 'NEGATIVE'
        score = 0.98

    print(f'Review: {review}')
    print(f' -> {label} ({score})\n')

# --------------------------------------------------
# DOCUMENT CLASSIFICATION
# --------------------------------------------------

document = 'The central bank raised interest rates to control rising inflation.'

candidate_labels = ['Politics', 'Economy', 'Sports', 'Technology']

keywords = {
    'Politics': ['election', 'government', 'minister', 'parliament'],
    'Economy': ['bank', 'inflation', 'interest', 'economy', 'rates'],
    'Sports': ['match', 'cricket', 'football', 'tournament'],
    'Technology': ['software', 'ai', 'computer', 'technology']
}

print('===== DOCUMENT CLASSIFICATION =====')
print('Document:', document)

doc_lower = document.lower()
scores = {}

for label in candidate_labels:
    score = sum(word in doc_lower for word in keywords[label])
    scores[label] = score

# Sort labels by score
sorted_labels = sorted(scores.items(), key=lambda x: x[1], reverse=True)

for label, score in sorted_labels:
    probability = round((score + 1) / (sum(scores.values()) + len(scores)), 3)
    print(f'{label}: {probability}')
```
